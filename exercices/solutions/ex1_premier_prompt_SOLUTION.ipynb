# Exercice 1 — Premier prompt — Solution

```polyglot-notebook
#r "nuget: Microsoft.SemanticKernel, 1.74.0"
#r "nuget: Microsoft.SemanticKernel.Connectors.Ollama, 1.74.0-alpha"
#pragma warning disable SKEXP0001,SKEXP0070
using Microsoft.SemanticKernel; using Microsoft.SemanticKernel.ChatCompletion; using Microsoft.SemanticKernel.Connectors.Ollama;
var endpoint=new Uri("http://localhost:11434");
var k=Kernel.CreateBuilder().AddOllamaChatCompletion("gemma3:4b",endpoint).Build();
var chat=k.GetRequiredService<IChatCompletionService>();
var h=new ChatHistory(); h.AddUserMessage("Donne un synonyme de rapide.");
Console.WriteLine(await chat.GetChatMessageContentAsync(h,new OllamaPromptExecutionSettings{Temperature=0f},k));
Console.WriteLine(await chat.GetChatMessageContentAsync(h,new OllamaPromptExecutionSettings{Temperature=1f},k));
```
