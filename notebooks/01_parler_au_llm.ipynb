# Module 1 — Parler à un LLM local (le Kernel)

## Objectif
Envoyer une première requête à un modèle **local** (Ollama) via **Semantic Kernel**.

> Prérequis : `docker compose up -d` (Ollama sur http://localhost:11434).

### Concept
Le **Kernel** de Semantic Kernel = le moteur qui héberge le modèle et exécute nos requêtes. On lui confie un *prompt* ; il renvoie une *complétion*.

### Tips
> Le 1er appel est lent (Ollama charge le modèle en mémoire). Prévenir l'auditoire — sinon ils croient que ça plante.

```polyglot-notebook
#r "nuget: Microsoft.SemanticKernel, 1.74.0"
#r "nuget: Microsoft.SemanticKernel.Connectors.Ollama, 1.74.0-alpha"
#pragma warning disable SKEXP0001, SKEXP0070
using Microsoft.SemanticKernel;
using Microsoft.SemanticKernel.ChatCompletion;
using Microsoft.SemanticKernel.Connectors.Ollama;
var endpoint = new Uri("http://localhost:11434");
```

```polyglot-notebook
var kernel = Kernel.CreateBuilder().AddOllamaChatCompletion("gemma3:4b", endpoint).Build();

var r = await kernel.InvokePromptAsync("Explique le RAG en une phrase simple.");
Console.WriteLine(r);
```

**À observer** : on n'a écrit ni HTTP, ni JSON — le connecteur Ollama s'en charge.

### Démo de l'hallucination (pour justifier le RAG)

```polyglot-notebook
// Question sur une donnée interne fictive : le modèle NE PEUT PAS savoir -> il risque d'inventer.
var halluc = await kernel.InvokePromptAsync(
    "Quel est le numero du ticket concernant la panne wifi de la salle 3 chez Cognitic ?");
Console.WriteLine(halluc);
```

**À observer** : la réponse est confiante mais **inventée**. C'est exactement le problème que le RAG va corriger.
