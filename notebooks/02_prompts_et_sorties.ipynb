# Module 2 — Prompts & sorties

## Objectif
Cadrer le comportement du modèle : rôle (system), style, température.

### Analogie
Le **system prompt** = les consignes du chef de projet ; le message **user** = la question du client.

### Tips
> `Temperature = 0` = réponses stables et reproductibles (idéal en démo et en RAG).

```polyglot-notebook
#r "nuget: Microsoft.SemanticKernel, 1.74.0"
#r "nuget: Microsoft.SemanticKernel.Connectors.Ollama, 1.74.0-alpha"
#pragma warning disable SKEXP0001, SKEXP0070
using Microsoft.SemanticKernel;
using Microsoft.SemanticKernel.ChatCompletion;
using Microsoft.SemanticKernel.Connectors.Ollama;
var endpoint = new Uri("http://localhost:11434");
```

```polyglot-notebook
var kernel = Kernel.CreateBuilder().AddOllamaChatCompletion("gemma3:4b", endpoint).Build();
var chat = kernel.GetRequiredService<IChatCompletionService>();

var history = new ChatHistory("Tu es un assistant support technique. Réponds en une phrase, sans blabla.");
history.AddUserMessage("Le Wi-Fi de la salle 3 est instable, que faire ?");

var settings = new OllamaPromptExecutionSettings { Temperature = 0f };
var rep = await chat.GetChatMessageContentAsync(history, settings, kernel);
Console.WriteLine(rep.Content);
```

**À observer** : change le system prompt (« réponds comme un pirate ») et relance : seul le rôle a changé.

## Récap
- `ChatHistory("system…")` + `AddUserMessage`.
- `Temperature = 0` pour la stabilité.
- Mais le modèle ne connaît toujours pas VOS données → Module 3.
