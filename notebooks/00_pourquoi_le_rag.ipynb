# Module 0 — Pourquoi le RAG ? (concepts)

## Le problème
Un LLM, seul, a deux limites majeures en entreprise :
1. **Il ne connaît pas VOS données** (procédures internes, tickets, contrats…). Elles ne sont pas dans son entraînement.
2. **Il hallucine** : quand il ne sait pas, il invente une réponse plausible mais fausse.

## L'idée du RAG
**RAG = Retrieval-Augmented Generation.** Au lieu de demander au modèle de « savoir », on lui **fournit les bons extraits** de vos documents au moment de répondre.

> **Analogie** : un étudiant qui passe un examen **avec ses fiches**. Il ne récite pas de mémoire (il inventerait) : il **cherche la bonne fiche**, puis rédige la réponse à partir d'elle.

## Le RAG en 3 gestes
1. **Ingérer** : découper vos documents et les stocker sous forme de **vecteurs** (embeddings) dans une base vectorielle.
2. **Récupérer** : pour une question, retrouver les **passages les plus proches par le sens**.
3. **Générer** : donner ces passages au LLM comme **contexte** et lui demander de répondre *uniquement* à partir d'eux (+ citer les sources).

## Pourquoi c'est intéressant (décideurs)
- **Vos données restent chez vous** (ici : 100 % local, aucune clé d'API).
- **Mise à jour facile** : ajouter un document = ré-ingérer, pas ré-entraîner.
- **Traçabilité** : la réponse cite ses sources → confiance & conformité.
- **Coût maîtrisé** : pas de fine-tuning, un petit modèle local suffit souvent.

## Quand le RAG n'est PAS la bonne réponse
- Question de **raisonnement pur** sans base documentaire → un bon prompt suffit.
- Données **structurées** (chiffres, agrégats) → une requête SQL est plus fiable.
- Besoin de **temps réel transactionnel** → un appel d'API, pas du RAG.

## Ce qu'on va construire aujourd'hui
Un **assistant support technique** qui répond à partir d'une base de connaissances interne (`data/support_kb/`), 100 % en local, en .NET avec **Semantic Kernel**.
